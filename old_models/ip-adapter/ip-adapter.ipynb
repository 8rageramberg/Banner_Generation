```python
from diffusers import AutoPipelineForText2Image
from diffusers.utils import load_image
import torch

# ✅ Use float32 to avoid black images on MPS
pipeline = AutoPipelineForText2Image.from_pretrained(
    "stabilityai/stable-diffusion-xl-base-1.0",
    torch_dtype=torch.float16
).to("mps")

# Load IP-Adapter
pipeline.load_ip_adapter(
    "h94/IP-Adapter", 
    subfolder="sdxl_models", 
    weight_name="ip-adapter_sdxl.bin"
)
pipeline.set_ip_adapter_scale(0.6)
```

```python
# Load your logo or image prompt
image = load_image("https://huggingface.co/datasets/huggingface/documentation-images/resolve/main/diffusers/ip_adapter_diner.png")

# Generator
generator = torch.Generator(device="mps").manual_seed(0)

# Run inference (slow on MPS, but float32 avoids errors)
images = pipeline(
    prompt="a polar bear sitting in a chair drinking a milkshake",
    ip_adapter_image=image,
    negative_prompt="deformed, ugly, wrong proportion, low res, bad anatomy, worst quality, low quality",
    num_inference_steps=5,  # reduce steps if needed
    generator=generator,
).images

images[0].save("output_ip_adapter_sdxl_mps.png")
```
